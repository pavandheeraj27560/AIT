```python
import heapq

def h(a, b):
  return abs(a[0]-b[0]) + abs(a[1]-b[1])
def astar(grid, start, goal):
  q = [(h(start, goal), 0, start, [start])]
  cost = {start: 0}

  while q:
    f, g, p, path = heapq.heappop(q)

    if p == goal:
      return path, g

    for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
      x, y = p[0] + dx, p[1] + dy
      neighbor = (x, y)
      if 0 <= x < len(grid) and 0 <= y < len(grid[0]):
        if grid[x][y] == 1:
            continue

        ng = g + 1
        if neighbor not in cost or ng < cost[neighbor]:
          cost[neighbor] = ng
          heapq.heappush(q, (ng + h(neighbor, goal), ng, neighbor, path + [neighbor]))

  return None, float('inf')
grid = [
    [0,0,0],
    [0,1,0],
    [0,0,0]
]
start = (0,0)
end = (2,2)

print("Input Grid:")
for row in grid:
    print(row)
print("Start:", start)
print("Goal:", end)

optimal_path, final_cost = astar(grid, start, end)
if optimal_path:
    print("Optimal path:", optimal_path)
    print("Cost:", final_cost)
else:
    print("No path found.")
```
