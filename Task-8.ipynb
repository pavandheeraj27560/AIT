```python
N = 4
board = [-1] * N

def safe(row, col):
  for i in range(row):
    if board[i] == col or abs(board[i] - col) == abs(i - row):
      return False
  return True

def print_board(board):
  for r in range(N):
    print("".join("Q" if board[r] == c else "." for c in range(N)))
  print()

def solve(row):
  if row == N:
    print("Solution found:")
    print_board(board)
    return True

  for col in range(N):
    if safe(row, col):
      board[row] = col
      if solve(row + 1):
        return True
      board[row] = -1
  return False
print(f"Attempting to solve N-Queens for N={N}")
if not solve(0):
  print("No solution exists for the given N.")
```
