```python
import math
board=['' for i in range(9)]
def print_board():
  for i in range (3):
    print(""+ board[i*3]+"|"+ board[i*3+1]+"|"+ board[i*3+2])
    if i<2:
      print("---+---+---")
def evaluate():
  wins=[(0,1,2),(3,4,5),(6,7,8),
        (0,3,6),(1,4,7),(2,5,8),
        (0,4,8),(2,4,6)
        ]
  for a,b,c in wins:
    if board[a]==board[b]==board[c] and board[a]!='':
      if board[a]=='0':
        return 1
      elif board[a]=='x':
        return -1
  return 0
def moves_left():
  return '' in board
def minimax(is_max):
  score=evaluate()
  if score==1 or score==-1:
    return score
  if not moves_left():
    return 0
  if is_max:
    best=-math.inf
    for i in range(9):
      if board[i]=='':
        board[i]='0'
        best=max(best,minimax(False))
        board[i]=''
    return best
  else:
    best=math.inf
    for i in range(9):
      if board[i]=='':
        board[i]='x'
        best=min(best,minimax(True))
        board[i]=''
    return best
def find_best_move():
  best_value=-math.inf
  best_move=-1
  for i in range(9):
    if board[i]=='':
      board[i]='0'
      move_value=minimax(False)
      board[i]=''
      if move_value > best_value:
        best_value=move_value
        best_move=i
  return best_move
print("TIC-TAC-TOE")
print("you=x,AI=0")
print("Positions:0 1 2 3 4 5 6 7 8\n")
print_board()
while moves_left() and evaluate()==0:
  try:
    move=int(input("Enter your move(0-8):"))
    if move<0 or move>8 or board[move]!='':
      print("Invalid move!")
      continue
    board[move]='x'
    print_board()
  except ValueError:
    print("Enter a number from 0 to 8")
    continue
  if evaluate()==-1:
    print("congratulations ! you won")
    break
  if not moves_left():
    print("It's a draw!")
    break
  print("AI is making a move...")
  ai_move=find_best_move()
  if ai_move != -1:
    board[ai_move]='0'
    print_board()
  else:
    print("AI could not find a move, it's a draw!")
    break
  if evaluate()==1:
    print("AI wins!")
    break
  if not moves_left():
    print("It's a draw!")
    break
```
